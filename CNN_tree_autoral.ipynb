```python
# notebook para treinar a nn e conferir com a pré-treinada do phylodeep

def load_data_wrapper():
    # Função semelhante ao load_data, porém, o y do training_data
    # agora é um vetor unitário, representando a classificação correta para x.
    # Para o validation_data e test_data, o y é semelhante a função load_data
    # Abrindo o npz_load com as árvores
    data = np.load('C:\\Users\\JPC\Documents\\MESTRADO\\Projeto\\gillespie-2\\deep_learning_dataset_refatorado.npz',allow_pickle=True)
    

    # Abrindo as info de classificação
    with open('C:\\Users\\JPC\\Documents\\MESTRADO\\Projeto\\gillespie-2\\input_classificacao.txt') as f:
        reader = csv.reader(f, delimiter="\t")
        input_class = list(reader)

    # Removendo o header das info de classificação
    input_class.pop(0)

    # Retornando a segunda coluna (coluna referente ao R_nought)
    # e em segunda transformando a lista em um np.array
    R_nought = [sublist[5] for sublist in input_class]
    R_nought = np.array(R_nought)

    transmission_rate = [sublist[1] for sublist in input_class]
    transmission_rate = np.array(transmission_rate)

    infectious_time = [sublist[6] for sublist in input_class]
    infectious_time = np.array(infectious_time)

    vectorized = []
    for n in range(0, len(transmission_rate)):
        vectorized.append([float(transmission_rate[n]), float(infectious_time[n])])

    #R_nought_vector = []
    ##np.array([1, 0]) if i == 1 else np.array([0, 1]) for i in R_nought
    #for i in R_nought:
    #    if i == '1':
    #        R_nought_vector.append([float(1)])
    #        #R_nought_vector.append([1, 0])
    #    else:
    #        #R_nought_vector.append([5, 1])
    #        R_nought_vector.append([float(5)])

    #R_nought_vector = np.array(R_nought_vector)



    training_data = list(zip(data['tree_data'][:20], R_nought[:20], transmission_rate[:20], infectious_time[:20], vectorized[:20]))
    

    #validation_data = list(zip(data['tree_data'][8000:9000], R_nought_vector[8000:9000]))

    test_data = list(zip(data['tree_data'][0:20], R_nought[0:20], transmission_rate[0:20], infectious_time[0:20], vectorized[0:20]))
    
    return training_data, test_data



# Criar o loop do mini-batch e confere os erros
# Tentar pelo tensorflow
```

```python
training_data, test_data = load_data_wrapper()
```

```python
x_train = [sublist[0] for sublist in training_data]
y_train = [float(sublist[3]) for sublist in training_data]
testar_formato =  [float(sublist[3]) for sublist in training_data]

x_train = np.asarray(x_train)
y_train = np.asarray(y_train)

#x_train_normalized = [(tree - min(tree))/(max(tree)-min(tree)) for tree in x_train]
#x_train_normalized = (x_train - min(x_train))/(max(x_train)-min(x_train))


#x_train_normalized = np.asarray(x_train_normalized)

#-------------------------------------------------------------------

x_test = [sublist[0] for sublist in test_data]
y_test = [float(sublist[3]) for sublist in test_data]

x_test = np.asarray(x_test)
y_test = np.asarray(y_test)
```

```python
x_train[1]
```

```python
modelo_derivado = tf.keras.models.Sequential([
    layers.Input(x_train[0].shape),
    layers.Reshape((201, 2)),
    layers.Conv1D(filters = 50, kernel_size = 3, activation='relu'),
    layers.Conv1D(filters = 50, kernel_size = 10, activation='relu'),
    layers.MaxPooling1D(pool_size=10),
    layers.Conv1D(filters = 80, kernel_size = 10, activation='relu'),
    layers.GlobalAveragePooling1D(),
    layers.Dense(64, activation='elu'),
    layers.Dense(32, activation='elu'),
    layers.Dense(16, activation='elu'),
    layers.Dense(8, activation='elu'),
    layers.Dense(2, activation='elu'),
    layers.Dense(1, activation='linear')
    ])

```

```python
modelo_derivado.compile(optimizer='adam', loss='MeanAbsolutePercentageError', metrics=['accuracy'])

modelo_derivado.summary()
```

```python
history = modelo_derivado.fit(x_train, y_train, epochs=10, batch_size=32, validation_split=0.2)
```
